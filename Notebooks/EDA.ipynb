```python
# Create a full range of dates from start to finish
full_range = pd.date_range(start=df.index.min(), end=df.index.max(), freq='D')

# Reindex the dataframe (this will create rows for missing days with NaN values)
df = df.reindex(full_range)
```

```python
df.info()
```

```python
# Calculate the difference between the actual number of rows and the number of days in the range
expected_days = (df.index.max() - df.index.min()).days + 1
actual_days = len(df)

print(f"Expected days: {expected_days}")
print(f"Actual rows: {actual_days}")

if expected_days == actual_days:
    print("✅ Your timeline is perfectly continuous. No need to reindex!")
else:
    print(f"⚠️ You are missing {expected_days - actual_days} days.")
```

```python
df['Children in HHS Care_lag_1'] = df['Children in HHS Care'].shift(1)
df['Children in HHS Care_lag_7'] = df['Children in HHS Care'].shift(7)
df['Children in HHS Care_lag_14'] = df['Children in HHS Care'].shift(14)
```

```python
df1=df
```

```python
# 2. Rolling Averages: The "Trend" over the last week/two weeks
df1['rolling_mean_7'] = df1['Children in HHS Care'].rolling(window=7).mean()
df1['rolling_var_7'] = df1['Children in HHS Care'].rolling(window=7).var()
```

```python
# Calculate Net Pressure
# (Inflow into HHS) minus (Outflow from HHS)
df1['HHS_Net_Pressure'] = (
    df1['Children transferred out of CBP custody'] - 
    df1['Children discharged from HHS Care']
)

#Create a 3-day rolling sum to see if the pressure is sustained
df1['HHS_Pressure_Trend'] = df1['HHS_Net_Pressure'].rolling(window=3).sum()
```

```python
df1['day_of_week'] = df1.index.dayofweek
df1['month'] = df1.index.month
df1['is_weekend'] = df1.index.dayofweek.isin([5, 6]).astype(int)

```
